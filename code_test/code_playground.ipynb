```python
import string

def vigenere_code(keyword_1:str, keyword_2:str, secret_message:str)->str:
    ''' 
    - keyword_1 is the word that makes the alphabet a 'keyed' alphabet
    - secret_message - the word we are trying to cipher
    - keyword_2 is the word we use to cipher the secret message
    
    '''
    alpha = string.ascii_lowercase
    keyword_1_alpha = list(keyword_1)
    for i in alpha:
        if not i in keyword_1:
            keyword_1_alpha.append(i)

    table_of_code = []
    for letter in range(len(keyword_1_alpha)):
        new_line = keyword_1_alpha[letter:len(keyword_1_alpha)] + keyword_1_alpha[0:letter]
        table_of_code.append(new_line)


    def until_equal_length(word1, word2):
        ''' trying to make word1 the same length as word2 - no amendments to word2'''
        a = len(word2) - len(word1)
        if a>0:
            new_word = word1 + word1[0:a]
        else: new_word = word1
        if len(new_word)<len(word2):
            new_word = until_equal_length(new_word, word2)

        assert len(new_word) >= len(word2)
        return new_word
    
    keyword_2 = until_equal_length(keyword_2, secret_message)

    list_of_keyword2_columns = []
    for keyword2_letter in keyword_2:
        for column in range(len(table_of_code)):
            if table_of_code[0][column] == keyword2_letter:
                list_of_keyword2_columns.append(column)
    list_of_secret_message_rows = []
    for secret_message_letter in secret_message:
        for row in range(len(table_of_code)):
            if table_of_code[row][0] == secret_message_letter:
                list_of_secret_message_rows.append(row)
    assert len(list_of_keyword2_columns) >= len(list_of_secret_message_rows)
    x = ''
    for i in range(len(list_of_secret_message_rows)):
        row = list_of_secret_message_rows[i]
        column = list_of_keyword2_columns[i]
        x += table_of_code[row][column]
    
    return x

# TODO add in allowance for spaces 
```
